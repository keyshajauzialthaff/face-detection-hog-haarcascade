```python
# Load OpenCV's built-in default face detector cascade file to avoid path issues
face_detector = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

image = cv2.imread('/content/drive/MyDrive/Colab Notebooks/adtt/dosenSTK.jpg')
image = cv2.resize(image, (800, 600))
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

detections = face_detector.detectMultiScale(
    image_gray,
    scaleFactor=1.25,
    minNeighbors=5
)

for (x, y, w, h) in detections:
    cv2.rectangle(image, (x, y), (x + w, y + h), (0, 255, 0), 5)

from google.colab.patches import cv2_imshow
cv2_imshow(image)

print("Faces detected:", len(detections))
```

```python
import dlib

image = cv2.imread('/content/drive/MyDrive/Colab Notebooks/adtt/dosenSTK.jpg')
image = cv2.resize(image, (800, 600))
cv2_imshow(image)

face_detector_hog = dlib.get_frontal_face_detector()
detections = face_detector_hog(image, 1)
detections

for face in detections:
    l, t, r, b = face.left(), face.top(), face.right(), face.bottom()
    cv2.rectangle(image, (l, t), (r, b), (0, 255, 255), 2)

cv2_imshow(image)
len(detections)
```
